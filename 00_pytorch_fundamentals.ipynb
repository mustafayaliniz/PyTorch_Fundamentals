### Matris Çarpımı

Yapay sinir ağlarında ve derin öğrenmede çarpma işlemini gerçekleştirmenin iki ana yolu vardır

1. Eleman bazında çarpma
2. Matris çarpımı (nokta çarpımı)
3. ![matris çarpımı](images/matrix_multiplication.jpg)

Matris çarpımının etmesi gereken iki ana kural vardır:
1. **İç boyutlar** eşleşmelidir:
* `(3, 2) @ (3, 2)` çalışmaz
* `(2, 3) @ (3, 2)` çalışır
* `(3, 2) @ (2, 3)` çalışır
2. Sonuç matrisinin şekli **dış boyutlar** olacaktır:
* `(2, 3) @ (3, 2) -> (2, 2)`
* `(3, 2) @ (2, 3) -> (3, 3)`

```python
# Eleman bazında çarpma
print(tensor, "*", tensor)
print(f'Eşittir: {tensor * tensor}')
```

```python
# Matris çarpımı
torch.matmul(tensor, tensor)
```

### Derin öğrenmedeki en yaygın hatalardan biri: şekil hataları

```python
# Matris çarpımı için şekiller
tensor_A = torch.tensor([[1, 2],
                        [3, 4],
                        [5, 6]])
tensor_B = torch.tensor([[7, 8],
                        [9, 10],
                        [11, 12]])
#torch.mm(tensor_A, tensor_B) # torch.mm, torch.matmul ile aynıdır
torch.matmul(tensor_A, tensor_B)
```

```python
tensor_A.shape, tensor_B.shape
```

Tensor şekil sorunlarımızı düzeltmek için, tensor'lardan birinin şeklini **transpose** kullanarak değiştirebiliriz.
Bir **transpose**, verilen bir tensor'un eksenlerini veya boyutlarını değiştirir.

```python
tensor_B, tensor_B.shape
```

```python
tensor_B.T, tensor_B.T.shape
```

```python
# Matris çarpımı, tensor_B transpoze edildiğinde çalışır
print(f"Orijinal şekiller: tensor_A= {tensor_A.shape} , tensor_B= {tensor_B.shape}")
print(f"Yeni şekiller: tensor_A= {tensor_A.shape} , tensor_B.T= {tensor_B.T.shape}")
print(f"Çarpma: {tensor_A.shape} * {tensor_B.T.shape}")
print("Çıktı:\n")
output = torch.matmul(tensor_A, tensor_B.T)
print(output)
```

## Reshaping, stacking, squeezing and unsqueezing tensors

* Reshaping - bir giriş tensor'unu tanımlı bir şekle yeniden şekillendirir
* View - belirli bir şekle sahip bir giriş tensor'unun görünümünü döndürür ancak orijinal tensor ile aynı belleği korur
* Stacking - birden fazla tensor'u üst üste (vstack) veya yan yana (hstack) birleştirir
* Squeeze - bir tensor'dan tüm `1` boyutlarını kaldırır
* Unsqueeze - hedef tensor'a bir `1` boyutu ekler
* Permute - girişin boyutlarını belirli bir şekilde permute edilmiş (değiştirilmiş) bir görünümünü döndürür

```python
# Bir tensor oluştur
import torch 
x = torch.arange(1., 10.)
x, x.shape
```

```python
# Ekstra bir boyut ekle
x_reshaped = x.reshape(1, 9)
x_reshaped, x_reshaped.shape
```

```python
# Görünümü değiştir
z = x.view(1, 9)
z, z.shape
```

```python
# z'yi değiştirmek x'i değiştirir (çünkü bir tensor'un görünümü orijinal giriş ile aynı belleği paylaşır)
z[:, 0] = 5
z, x
```

```python
# Tensor'ları üst üste yığ
x_stacked = torch.stack([x, x, x, x], dim=0)
x_stacked
```

```python
# torch.squeeze() - hedef tensor'dan tüm tek boyutları kaldırır
print(f"Önceki tensor: {x_reshaped}")
print(f"Önceki şekil: {x_reshaped.shape}")

# x_reshaped'den ekstra boyutları kaldır
x_squeezed = x_reshaped.squeeze()
print(f"\nYeni tensor: {x_squeezed}")
print(f"Yeni şekil: {x_squeezed.shape}")
```

```python
# torch.unsqueeze() - belirli bir boyutta hedef tensor'a tek bir boyut ekler
print(f"Önceki hedef: {x_squeezed}")
print(f"Önceki şekil: {x_squeezed.shape}")

# Unsqueeze ile ekstra bir boyut ekle
x_unsqueezed = x_squeezed.unsqueeze(dim=0)
print(f"\nYeni tensor: {x_unsqueezed}")
print(f"Yeni şekil: {x_unsqueezed.shape}")
```

```python
# torch.permute - hedef tensor'un boyutlarını belirli bir sırayla yeniden düzenler
x_original = torch.rand(size=(224, 224, 3))

# Orijinal tensor'un eksen (veya boyut) sırasını yeniden düzenlemek için permute et
x_permuted = x_original.permute(2,0,1)

print(f"Önceki şekil: {x_original.shape}")
print(f"Yeni şekil: {x_permuted.shape}")
```

## PyTorch tensor'ları & NumPy

NumPy, popüler bir bilimsel Python sayısal hesaplama kütüphanesidir, ve bu nedenle, PyTorch onunla etkileşim kurma işlevselliğine sahiptir.
* NumPy'daki veri, PyTorch tensor'una dönüştürülmek isteniyor -> `torch.from_numpy(ndarray)`
* PyTorch tensor'u -> NumPy -> `torch.Tensor.numpy()`

```python
# NumPy dizisinden tensor'a
import torch
import numpy as np
array = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array)
array, tensor
```

```python
# Diziyi değiştir, tensor'u koru
array = array + 1
array, tensor
```

```python
# Tensor'dan NumPy dizisine
tensor = torch.ones(7) # float32 veri tipinde birlerden oluşan bir tensor oluştur
numpy_tensor = tensor.numpy() # veri tipi float32 olacaktır, aksi belirtilmedikçe
tensor, numpy_tensor
```

```python
# Tensor'u değiştir, diziyi aynı tut
tensor = tensor + 1
tensor, numpy_tensor
```

## Tekrarlanabilirlik (rastgeleliği rastgeleden çıkarmaya çalışmak)

Kısaca yapay sinir ağı öğrenmesi:

`rastgele sayılarla başla -> tensor işlemleri -> daha iyisini yapmaya çalış (tekrar tekrar tekrar)`

Yapay sinir ağlarında ve PyTorch'ta rastgeleliği azaltmak için **rastgele tohum** kavramı gelir.

Temelde rastgele tohum, rastgeleliği "tatlandırır".

```python
import torch 

# İki rastgele tensor oluştur
random_tensor_A = torch.rand(3,4)
random_tensor_B = torch.rand(3,4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# Rastgele ama tekrarlanabilir tensor'lar yapalım
import torch 

# Rastgele bir tohum ayarla
torch.manual_seed(42)
random_tensor_C = torch.rand(3,4)
torch.manual_seed(42)
random_tensor_D = torch.rand(3,4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D)
```

Tekrarlanabilirlik için ek kaynaklar:
